```python
import pandas as pd
df = pd.read_csv ('https://raw.githubusercontent.com/CunyLaguardiaDataAnalytics/datasets/refs/heads/master/311_Service_Requests_from_2019May.csv')
```

```python
#How many colums? 21
#How many rows? 69637
df.shape
```

```python
#Summary statistics
df["Agency"].describe()
```

```python
#Visualizations
df.groupby(["Status"]) ["Unique Key"].count().sort_values(ascending=False).plot(kind= "bar",
                                                                                ylabel= "Count of Compaints", title= "Count of Complaints by Status")
```

```python
df["Complaint Type"].value_counts().head(5).plot(kind= "bar", ylabel= "Count of Complaints", title= "Top 5 Complaint Types", color="orange")
```

```python
#Which borough has the most complaints? Brooklyn
df.groupby(['Borough']) ['Unique Key'].count(). sort_values(ascending=False)
```

```python
#Plot of counts
df[df["Complaint Type"] =='Noise'].groupby(['Borough']) ['Unique Key'].count(). sort_values(ascending=False).plot(kind='bar',
                                                                                                                  ylabel= 'Count of Complaints', title = "Count of Noise Complaints by Borough", color = "green")
```

```python
#More interesting info
df[df["Status"].isin(['Started', 'Email Sent'])].groupby(['Status']) ['Unique Key'].count(). sort_values(ascending=False)
```

```python
df[df["Status"].isin(['Started', 'Email Sent'])].groupby(['Status']) ['Unique Key'].count(). sort_values(ascending=False).plot(kind= 'bar',
                                                                                                                               ylabel= 'Count of Selected Status', title= 'Complaints by Selected Status')
```
